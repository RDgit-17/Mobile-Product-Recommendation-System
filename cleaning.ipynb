## Data cleaning

```python
df.info()
```

The following columns don't add any value:
- review_id
- customer_name
- language (not required when there is country column present)

So, duplicates are verified and these columns are dropped.

```python
print(df.duplicated().sum())
print(df.drop(columns="review_id").duplicated().sum())
print(df.duplicated(subset=['customer_name','age','country','language','brand','model','source','rating']).sum())
```

```python
df = df.drop(columns=["customer_name", "review_id", "language"])
```

- The price_local must be cleaned to be proper decimal numbers.
- The rows where both price_local and price_usd values are null are dropped. 
- After that null price values can be imputed using formula.

```python
# --------------- cleaning price_local values ------------------
df["price_local"] = (df["price_local"].str.replace(",", "", regex=False).str.extract(r"(\d+(?:\.\d+)?)")[0].astype(float))

# --------------- removing rows where both price values are empty --------------
to_remove = df.query("price_local.isnull() and price_usd.isnull()")
df.drop(to_remove.index, inplace=True)

# ------------- wherever single prices are missing, formula is used to fill -------------
df["price_usd"] = df["price_usd"].fillna(df["price_local"] / df["exchange_rate_to_usd"])
df["price_local"] = df["price_local"].fillna(df["price_usd"] * df["exchange_rate_to_usd"])

df[["price_usd", "price_local"]] = df[["price_usd", "price_local"]].round(2)
```

Rating value should not be imputed, so dropping rows where rating is null

```python
df.dropna(subset=['rating'], inplace=True)
```

```python
df.info()
```

```python
print(df['sentiment'].value_counts())
print()
print(df['source'].value_counts())
```

```python
df.groupby('rating')['sentiment'].value_counts()
```

**Sentiment column:**

- There was no consistency in sentiment column. 
- Sentiments are not mapped clearly to the rating, so it is difficult to determine based on what sentiment is being decided.
- It is also not going to add value to the recommendation system as it is more like an inference based on the rating, and not an individual feature.
- So sentiment column can be dropped.

**Source column:**
- Source column has null entries.
- Filling them with the mode will increase the weightage of one source to extreme level, as all the sources have equal weightage.
- Source can be added as a part of the recommendation system, so not dropping it.
- Null entries in the column could be filled with 'Unknown' 

```python
df.drop(columns='sentiment',inplace = True) 
df["source"] = df["source"].fillna("Unknown")
```

```python
for i in text_columns:
  df[i] = df[i].str.strip()

df['review_date'] = pd.to_datetime(df['review_date'])

df['verified_purchase'] = df['verified_purchase'].astype(int)

df['rating'] = df['rating'].astype(int)
```
